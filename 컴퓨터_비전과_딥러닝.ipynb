## 📝 4.1 에지 검출_ 소벨 에지 검출

- 에지 검출 알고리즘은 물체 내부는 명암이 서서히 변하고 물체 경계는 명암이 급격히 변하는 특성을 이용.
    - 밝기 변화가 없는 구간: 미분하면 기울기가 0 에 가까움.
    - 에지 구간(= 밝기가 급변하는 구간): 미분하면 기울기가 큰 값



###🔹 소벨 에지 검출(Sobel)

- 영상의 가로(**x축 커널**로 세로 선들 찾아) +  세로(**y축 커널**로 가로 선들 찾아) 방향으로 각각 미분----> 최종적인 에지(**x축 + y축**)를 찾아내는 방식
    - 미분값이 크다 =  기울기가 가파르다 = 밝기 변화가 급격하다 = 에지가 확실하다

    - 미분값이 작다 = 기울기가 완만하다 = 밝기 변화가 서서히 일어난다 = 에지가 흐릿하거나 없다

## 📝 4.2 캐니 에지

###🔹 캐니 에지(Canny edge)
- 최소 오류율, 위치 정확도, 한 두께라는 세 가지 기준으로 목적 함수 정의

    4단계:
    1. 노이즈 제거 - blur
    2. 에지 기울기 사용 - 소벨 연산(x축, y축 -> 강도 & 방향), 미분 차이 계산
    3. 비최대 억제 - 날카롭게 만들기 위해 양옆을 깎아냄
    4. 이중 임계값 및 히스테리시스 - 정확하게 선별 위해 가짜 선 통째로 지움


## 📝 4.3. 직선 검출

###🔹 허프 변환(Hough transform)
    - 직선 후보가 아니라 원 후보에게 투표를 하는 것.
    - 에지 픽셀들이 후보들에게 투표(Voting)를 하고, 누적 배열(Accumulator)에서 가장 많은 표를 얻은 후보가 직선으로 당선되는 원리
        - 점들의 인기 투표로 진짜 직선,원 찾기
        - 모든 에지 점이 자기를 지날 수 있는 직선,원 후보들에 투표
        - 표가 가장 많이 겹치는 후보가 진짜 직선,원으로 당선
        - 한계 극복: 직선에서는 거리와 각도를 사용함, 원에서는 중심점과 반지름


###🔹 RANSAC(RANdom SAmple Consensus)
- 이상치(Outlier)가 섞인 데이터에서 진짜 데이터(Inlier)만 골라내어 모델을 세우는 방법
    - 랜덤하게 두 점을 선택
    - 선택한 점으로 직선을 그어봄
    - 직선 근처에 점(인라이너)가 몇 개 있는지 확인
    - 가장 많은 점이 지지하는 모델을 최종 선택

        - 반복 횟수가 많을수록 진짜 직선을 찾을 확률 높아짐.
        - 최소자승법(Least Squares)(선 긋는 기술)이 아웃라이어 한두 개에 휘둘려 모델이 망가지는 단점을 완벽히 극복함.

## 📝 4.4. 영역 분할



###🔹 단순 영역 분할
    - 이진화 알고리즘(오츄 알고리즘, 군집화 알고리즘)
        > 픽셀 값의 분포를 분석해 배경과 물체를 0과 1로 나눔
    - 워터 셰드(Watershed)
        > 영상을 하나의 지형으로 간주하고 물체를 분할하는 기법

        > 에지(경계)를 높은 산맥으로 간주하고 낮은 분지부터 물을 채워 영역을 나누는 방식


###🔹슈퍼 화소 분할
    - SLIC(Simple Linear Iterative Clustering):
        > 색상(R,G,B)과 좌표(x,y)가 유사한 픽셀들을 하나의 덩어리로 묶음
        
        > 픽셀로 분석하는 대신, 의미 있는 뭉치(슈퍼픽셀) 단위로 분석

###🔹최적화 분할
- 이전까지는 지역적 명암 변화만 살펴 비슷한 명암을 가진 부분에선 경계가 형성되지 않아 물체를 분할하는 문제가 발생.
- 지역적 명암 변화를 보되 전역적 정보를 같이 고려


- 영상의 그래프 표현
    > 픽셀을 노드로, 관계를 에지로 표현
    > 거리와 유사도: 두 픽셀의 값(색상, 밝기 등) 차이가 적으면 **"거리가 가깝다"**고 하며, 유사도가 높아 에지의 연결이 강력해짐.

- 정규화 절단 알고리즘
    > 정규화 절단은 화소를 노드로 취함

    > 색상(R,G,B)과 위치(x,y)를 결합한 5차원 벡터를 사용
    
    > 유사도를 에지 두께(가중치)로 사용
    
    > cut은 영역 분할의 좋은 정도를 측정해주는 목적 함수
    
    > ncut 함수를 목적 함수로 사용하면 분할을 최적화 문제로 풀 수 있음(ncut 작다 = 좋은 분할)

## 📝 4.5. 대화식 분할

###🔹능동 외곽선
- 사용자가 물체 내부에 초기 곡선을 지정하면 곡선을 점점 확장하면서 물체 외곽선으로 접근하는 방법, 스네이크라는 별명 얻음
- 곡선 결정하는 3대 에너지를 최소화하는 방향으로 곡선이 움직임
    > 3대 에너지: 영상에너지, 내부에너지, 도메인에너지(외부에너지)의 합(전체에너지)으로 표현

###🔹GrabCut
- 사용자가 물체(전경)와 배경을 대략적으로 지정하면, 컴퓨터가 반복적 최적화를 통해 정밀하게 경계를 찾아내는 기법.
    > 사용자가 붓으로 물체와 배경을 초기 지정함
   
    > 파란색은 물체, 빨간색은 배경
   
    > 물체 히스토그램과 배경 히스토그램을 만듦
   
    > 새로운 정보로 히스토그램을 다시 만들고 물체 영역과 배경 영역을 갱신하는 과정 반복
   
    > 영역이 거의 변하지 않으면 수렴한 것으로 간주하고 멈춤

- GrabCut의 한계: 색상 유사성으로 인한 오인(물체와 배경의 색상 히스토그램이 겹칠 경우, 경계를 정확히 구분하지 못하고 배경을 물체로(혹은 반대로) 오인하는 경우가 발생.)

## 📝 4.6. 영역 특징

###🔹불변성(Invariant)
- 변환을 해도 특징의 값이 변하지 않음
    > 면적: 회전에는 불변 O/ 축소에는 불변 X

###🔹등변성(Equivariant)
- 변환에 따라 값이 변함
        > 면적: 축소에는 등변 O/ 회전에는 등변 X

###🔹텍스처 특징
- LBP(Local binary pattern)
    > 주위 8개 픽셀의 명암차이를 0,1로 표시(어두우면 0/ 밝으면 1)

    > 중심 화소와 주위 화소의 명암 값을 비교해서 텍스처 측정, 256차원 특징 벡터

- LTP(Local ternary pattern)
    > 작은 명암 변화에 민감한 LBP 단점 개선(0과 1로 나누기 때문에 노이즈에 민감함)
    > 임계값을 두어 3진수로 표시-->
    
        '중심 - 임계값' 보다 작으면 '-1'
        '중심 + 임계값' 보다 크면 '+1'
        '중심' 유사하면 '0'
    
    >  LBP보다 노이즈에 강하며, 평탄한 영역에서 더 안정적인 성능을 보임, 512차원 특징 벡터
